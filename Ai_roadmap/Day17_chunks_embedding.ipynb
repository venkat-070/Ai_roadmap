```python
text = """
Artificial intelligence is transforming the world.
Machine learning is a subset of AI that enables systems to learn from data.
Deep learning uses neural networks with many layers.
Natural language processing helps computers understand human language.
Computer vision allows machines to interpret visual information.
Large language models are trained on vast amounts of text data.
These models can generate human-like text responses.
RAG stands for Retrieval Augmented Generation.
It combines search with language generation.
This makes AI systems more accurate and reliable.
"""

chunk_size = 100
overlap = 20

chunks = []
for i in range(0, len(text), chunk_size - overlap):
    chunk = text[i:i+chunk_size]
    chunks.append(chunk)

for idx, chunk in enumerate(chunks):
    print(f"Chunk {idx+1}:")
    print(chunk)
    print("---")
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
chunks = [
    "Refund requests must be made within 30 days of purchase.",
    "Shipping takes 5-7 business days for standard delivery.",
    "Contact our support team at support@example.com for help.",
    "Products can be returned in original packaging only.",
    "Express shipping is available for an additional fee."
]

question = "How do I return a product?"

Chunks_embedding = model.encode(chunks)
question_embedding = model.encode(question)

print(Chunks_embedding.shape)
print(question_embedding.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

scores = cosine_similarity([question_embedding],Chunks_embedding)[0]

for i,score in enumerate(scores):
  print(f"chunk {i+1}:,{score:.4f}-{chunks[i]}")
```
